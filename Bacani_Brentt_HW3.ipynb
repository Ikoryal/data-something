```python
import pandas as pd

df = pd.read_csv('titanic.csv')
embarkMode = df['Embarked'].mode()[0]
print(f"Most frequent 'Embarked' value: {embarkMode}")
df['Embarked'] = df['Embarked'].fillna(embarkMode)
ageMean = df['Age'].mean()
print(f"Median age of passengers: {ageMean:.2f}")
df['Age'] = df['Age'].fillna(ageMean)
df['Embarked']
```

```python
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
familySizeMin = df['FamilySize'].min()
familySizeMax = df['FamilySize'].max()
familySizeMean = df['FamilySize'].mean()
df['IsAlone'] = (df['FamilySize'] == 1).astype(int)
isAlone = df['IsAlone'].sum()
print(f"Minimum family size: {familySizeMin}")
print(f"Maxium family size: {familySizeMax}")
print(f"Mean family size: {familySizeMean:.2f}")
print(f"Amount of people travelling alone: {isAlone}")
```

```python
dfFemale = df[df['Sex'] == 'female']
dfMale = df[df['Sex'] == 'male']
dfSixteen = df[df['Age'] < 16]

dfFemaleSurvivalRate = dfFemale['Survived'].mean()
dfMaleSurvivalRate = dfMale['Survived'].mean()
dfSixteenSurvivalRate = dfSixteen['Survived'].mean()
print(f"Percent of female survivors: {dfFemaleSurvivalRate * 100:.2f}%")
print(f"Percent of male survivors: {dfMaleSurvivalRate * 100:.2f}%")
print(f"Percent of survivors under 16: {dfSixteenSurvivalRate * 100:.2f}%")

sexAndPClassSurvival = df.groupby(['Sex', 'Pclass'])['Survived'].mean()
print(sexAndPClassSurvival.round(2))
print("On average, the people who survived the most are females with a higher Pclass. On the other hand, males with a lower Pclass had the lowest. This means that you had a higher survival rate if you are female and/or of a higher Pclass")
```

```python
ticketGroupSize = df.groupby('Ticket')['Ticket'].transform('count')
df['farePerPerson'] = df['Fare'] / ticketGroupSize

avgFarePrice = df.groupby('Pclass')['farePerPerson'].mean()
avgFarePrice.round(2)
```
