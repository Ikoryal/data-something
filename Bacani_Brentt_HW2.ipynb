```python
import numpy as np

def get_grade_letter(finalGrade):
    if finalGrade >= 90.0:
        return "A"
    elif 80.0 <= finalGrade < 90:
        return "B"
    elif 70.0 <= finalGrade < 80:
        return "C"
    elif 60.0 <= finalGrade < 70:
        return "D"
    else:
        return "F"


names = []
scores = []
weights = np.array([0.05, 0.15, 0.20, 0.20, 0.40])

for i in range(10):
    name = input("Input the name of the student: ")
    names.append(name)

    studentScores = []
    for j in range(5):
        score = float(input(f"Input the score of test #{j+1}: "))
        studentScores.append(score)
    scores.append(studentScores)

names = np.array(names)
scores = np.array(scores)

finalGrades = scores @ weights

print("\nFinal grades:")
for name, finalGrade in zip(names, finalGrades):
    print(f"{name}: {finalGrade:.2f} -> " + get_grade_letter(finalGrade))
```
